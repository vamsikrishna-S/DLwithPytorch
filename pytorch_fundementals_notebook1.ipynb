```python
# create a cpu tensor multiplication and gpu tensor multiplication and compare time
import time

# create a large tensor
tensor = torch.randn(1000, 1000)

# CPU multiplication
start = time.time()
cpu_result = tensor @ tensor
end = time.time()
print(f"CPU time: {end - start}")

# GPU multiplication (if available)
if torch.cuda.is_available():
    tensor = tensor.to("cuda")
    start = time.time()
    gpu_result = tensor @ tensor
    end = time.time()
    print(f"GPU time: {end - start}")
    # Move the result back to CPU for comparison if needed
    gpu_result = gpu_result.to("cpu")
    # Compare CPU and GPU results
    print(f"Results are equal: {torch.allclose(cpu_result, gpu_result)}")
```

```python
### another example of a matrix multiplication
#all below approaches are for matrix multiplication
tensor_A = torch.tensor([[1,2],[3,4],[5,6]])
tensor_B = torch.tensor([[7,8,9],[10,11,12]])
print(tensor_A @ tensor_B)
print(torch.matmul(tensor_A, tensor_B))
print(torch.mm(tensor_A, tensor_B))
```

```python
#### basic matrix multiplication errors
tensor_C = torch.tensor([[1,2,3],[4,5,6]])
tensor_D = torch.tensor([[7,8],[9,10]])
# This will raise an error because the inner dimensions do not match
try:
    print(tensor_C @ tensor_D)
except RuntimeError as e:
    print(f"Error: {e}")
```

```python
#### matrix transpose
# create a matrix
matrix = torch.tensor([[1, 2], [3, 4]])
# transpose the matrix
transposed_matrix = matrix.T
print(f"Transposed matrix:\n{transposed_matrix}")
print(f"Original matrix:\n{matrix}")
```

```python
#### FINDING MIN , MAX , MEAN and SUM from a Tensor -- Tensor Aggregation
### create a 2D tensor with values from 0 to 90 with step 10 
tensor = torch.arange(0,100,10).reshape(2,5)
print(f"Tensor:\n{tensor}")
print(f"Min: {tensor.min()}")
print(f"Max: {tensor.max()}")
print(f"Mean: {tensor.float().mean()}")
print(f"Sum: {tensor.sum()}")


#### if we want to find the min, max, mean, and sum along a specific dimension
print(f"Min along dim 0: {tensor.min(dim=0).values}")
print(f"Max along dim 0: {tensor.max(dim=0).values}")
print(f"Mean along dim 0: {tensor.float().mean(dim=0)}")
print(f"Sum along dim 0: {tensor.sum(dim=0)}")

print(f"Min along dim 1: {tensor.min(dim=1).values}")
print(f"Max along dim 1: {tensor.max(dim=1).values}")
print(f"Mean along dim 1: {tensor.float().mean(dim=1)}")
print(f"Sum along dim 1: {tensor.sum(dim=1)}")
```
